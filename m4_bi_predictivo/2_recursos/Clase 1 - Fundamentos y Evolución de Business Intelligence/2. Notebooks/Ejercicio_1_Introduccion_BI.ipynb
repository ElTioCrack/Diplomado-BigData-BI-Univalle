## Paso 0: Crear dataset

Como estamos en clase, generaremos un dataset simple directamente en el notebook.
Esto evita depender de archivos externos.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Crear dataset sintético
num_registros = 300

df = pd.DataFrame({
    'cliente_id': np.random.randint(1, 101, num_registros),
    'canal': np.random.choice(['Tienda', 'Online'], num_registros),
    'tipo_cliente': np.random.choice(['Nuevo', 'Recurrente'], num_registros),
    'importe': np.random.randint(10, 200, num_registros)
})

# Generar variable de recompra (simplificada)
df['recompra'] = (df['tipo_cliente'] == 'Recurrente').astype(int)

```

## Paso 5: Buscar insights

Aquí buscamos significado:
- ¿Qué canal vende más?
- ¿Qué tipo de cliente es más valioso?

```python
df.groupby('canal')['importe'].mean()
```

```python
df.groupby('tipo_cliente')['recompra'].mean()
```

## Paso 6: Decisión

Ahora responde:
- ¿Qué canal priorizarías?
- ¿Qué cliente es más estratégico?
- ¿Qué datos adicionales pedirías?

El objetivo no es la respuesta perfecta, sino una interpretación coherente.

```python
# 1. ¿Qué canal priorizarías?
canal_stats = df.groupby('canal')['importe'].sum()
canal_priorizar = canal_stats.idxmax()
print(f"Priorizar canal: {canal_priorizar} (Ingresos: {canal_stats[canal_priorizar]} EUR)")

# 2. ¿Qué cliente es más estratégico?
cliente_stats = df.groupby('tipo_cliente')['recompra'].mean()
cliente_estrategico = cliente_stats.idxmax()
print(f"Cliente estratégico: {cliente_estrategico} (Tasa recompra: {cliente_stats[cliente_estrategico]*100}%)")

# 3. ¿Qué datos adicionales pedirías?
# 3. ¿Qué datos adicionales pedirías?
datos_extra = [
    "CAC (Coste de Adquisición): Para calcular la rentabilidad real de cada canal.",
    "Fechas de compra: Para medir la frecuencia y retención mediante análisis RFM.",
    "Margen neto: Para saber qué productos o canales dejan mayor beneficio real."
]
print("Datos adicionales necesarios:\n" + "\n".join(f"- {d}" for d in datos_extra))
```

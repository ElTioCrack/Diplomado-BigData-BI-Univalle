```python
import pandas as pd
import numpy as np

# 1. Configuración de datos sintéticos con lógica de negocio extendida
np.random.seed(42)
n = 300

# Base inicial con variables demográficas y transaccionales añadidas
df_ext = pd.DataFrame({
    'cliente_id': np.random.randint(1, 101, n),
    'edad': np.random.randint(18, 65, n),                             # Nueva columna: Segmentación etaria
    'canal': np.random.choice(['Tienda', 'Online'], n, p=[0.4, 0.6]),
    'tipo_cliente': np.random.choice(['Nuevo', 'Recurrente'], n, p=[0.5, 0.5]),
    'categoria_prod': np.random.choice(['Electro', 'Moda', 'Hogar'], n), # Nueva columna: Categoría
    'importe_bruto': np.random.randint(10, 200, n),
    'cantidad_items': np.random.randint(1, 6, n)                      # Nueva columna: Volumen por ticket
})

# Lógica condicionada para evitar datos absurdos y crear patrones reales
df_ext['descuento'] = np.where(df_ext['tipo_cliente'] == 'Recurrente', np.random.randint(5, 35, n), np.random.randint(0, 15, n))
df_ext['costo_envio'] = np.where(df_ext['canal'] == 'Online', np.random.randint(5, 20, n), 0)
df_ext['satisfaccion'] = np.where(df_ext['tipo_cliente'] == 'Recurrente', np.random.randint(3, 6, n), np.random.randint(1, 5, n))
df_ext['devolucion'] = np.where(df_ext['canal'] == 'Online', np.random.choice([0, 1], n, p=[0.85, 0.15]), np.random.choice([0, 1], n, p=[0.95, 0.05])) # Nueva columna: Tasa de retorno

# 2. Ingeniería de características (Métricas financieras combinadas)
df_ext['importe_neto'] = df_ext['importe_bruto'] - df_ext['descuento']
df_ext['gasto_total_cliente'] = df_ext['importe_neto'] + df_ext['costo_envio']
df_ext['precio_medio_item'] = df_ext['importe_bruto'] / df_ext['cantidad_items'] # Nueva columna: Valor unitario

# 3. Extracción de 3 Insights estratégicos multivariable
print("=== ANALÍTICA COMERCIAL: 3 INSIGHTS AVANZADOS ===\n")

# Insight 1: Preferencia de categoría y ticket promedio según rango de edad
df_ext['rango_edad'] = pd.cut(df_ext['edad'], bins=[17, 35, 50, 65], labels=['Jóvenes', 'Adultos', 'Seniors'])
ins_1 = df_ext.groupby(['rango_edad', 'categoria_prod'])['gasto_total_cliente'].mean().unstack()
print(f"1. Gasto Promedio por Rango de Edad y Categoría:\n{ins_1.round(2).to_string()}\n")

# Insight 2: Eficiencia operativa y pérdidas por devoluciones en canales digitales
ins_2 = df_ext.groupby('canal')['devolucion'].mean() * 100
print(f"2. Tasa de Devolución (%) por Canal de Venta:\n{ins_2.round(2).to_string()}\n")

# Insight 3: Rentabilidad real cruzando recurrencia y volumen de compra
ins_3 = df_ext.groupby('tipo_cliente').agg(
    ticket_neto_promedio=('importe_neto', 'mean'),
    items_por_compra=('cantidad_items', 'mean')
)
print(f"3. Comportamiento de Compra y Margen Neto por Tipo de Cliente:\n{ins_3.round(2).to_string()}")
```

## 1. Construcción del Dataset Base

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

np.random.seed(42)

# Dataset base original
df = pd.DataFrame({
    'cliente_id':   np.random.randint(1, 101, 300),
    'canal':        np.random.choice(['Tienda', 'Online'], 300),
    'tipo_cliente': np.random.choice(['Nuevo', 'Recurrente'], 300),
    'importe':      np.random.randint(10, 200, 300)
})

df.head()
```

---
## 2. Expansión del Dataset

Se añaden variables adicionales para simular un entorno de negocio más realista.

```python
# --- Nuevas columnas con np.random ---

# Porcentaje de descuento aplicado (0% – 30%)
df['descuento_aplicado'] = np.random.uniform(0, 0.30, 300).round(2)

# Costo de envío: solo aplica en canal Online; Tienda = 0
df['costo_envio'] = np.where(
    df['canal'] == 'Online',
    np.random.randint(5, 25, 300),
    0
)

# Puntuación de satisfacción del cliente (escala 1–5)
df['puntuacion_satisfaccion'] = np.random.randint(1, 6, 300)

df.head()
```

---
## 3. Métricas Calculadas

Se construyen indicadores derivados combinando las columnas existentes.

```python
# Importe neto tras aplicar el descuento
df['importe_neto'] = (df['importe'] * (1 - df['descuento_aplicado'])).round(2)

# Gasto total real del cliente (neto + envío)
df['gasto_total'] = (df['importe_neto'] + df['costo_envio']).round(2)

# Costo del descuento: cuánto "regala" la empresa por transacción
df['costo_descuento'] = (df['importe'] - df['importe_neto']).round(2)

print(f"Dataset final: {df.shape[0]} filas × {df.shape[1]} columnas")
df.head(10)
```

```python
# ============================================================================
# PALETA DE COLORES FIJA PARA TODOS LOS GRAFICOS
# ============================================================================
COLOR_TIENDA = '#b80144'      # Rojo Univalle
COLOR_ONLINE = '#e85d04'      # Naranja
COLOR_RECURRENTE = '#2b9348'  # Verde
COLOR_NUEVO = '#0077b6'       # Azul
COLOR_OTRO = '#818285'        # Gris

def get_color_canal(canal):
    if canal == 'Tienda':
        return COLOR_TIENDA
    elif canal == 'Online':
        return COLOR_ONLINE
    else:
        return COLOR_OTRO

def get_color_cliente(tipo):
    if tipo == 'Recurrente':
        return COLOR_RECURRENTE
    elif tipo == 'Nuevo':
        return COLOR_NUEVO
    else:
        return COLOR_OTRO
```

---
## 4. Insights de Negocio

> A continuación se presentan **3 descubrimientos clave** orientados a la toma de decisiones comerciales.

### Insight 1 — Rentabilidad neta por canal

**Pregunta:** ¿Qué canal genera más ingreso real después de descontar envíos y descuentos?

```python
rentabilidad_canal = df.groupby('canal').agg(
    ingreso_neto_total = ('importe_neto', 'sum'),
    costo_envio_total = ('costo_envio', 'sum'),
    costo_descuento_total = ('costo_descuento', 'sum'),
    transacciones = ('importe', 'count')
).assign(
    margen_operativo = lambda x: x['ingreso_neto_total'] - x['costo_envio_total']
).round(2)

print("\n" + "-"*50)
print("RENTABILIDAD POR CANAL")
print("-"*50)
display(rentabilidad_canal)

canal_ganador = rentabilidad_canal['margen_operativo'].idxmax()
print("\n" + "-"*50)
print("RESULTADO:")
print(f"Canal con mayor margen operativo: [{canal_ganador}]")
print("  Decision: priorizar inversion publicitaria en este canal.")
print("  Observacion: Online genera mas volumen, pero Tienda retiene mas margen")
print("  porque no tiene costo de envio.")
print("-"*50)

# Grafico con grid y colores por canal
plt.figure(figsize=(10,5))

colores = [get_color_canal(c) for c in rentabilidad_canal.index]
rentabilidad_canal['margen_operativo'].sort_values().plot(kind='barh', color=colores)

plt.title('Margen Operativo por Canal', fontsize=14, fontweight='bold')
plt.xlabel('Margen (euros)', fontsize=11)
plt.ylabel('Canal', fontsize=11)
plt.grid(axis='x', alpha=0.3, linestyle='--', color='#cccccc')

for i, v in enumerate(rentabilidad_canal['margen_operativo'].sort_values()):
    plt.text(v, i, f' {v:,.0f} €', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Insight 2 — Satisfacción vs. Gasto por tipo de cliente

**Pregunta:** ¿Los clientes que más gastan también son los más satisfechos?

```python
satisfaccion_gasto = df.groupby('tipo_cliente').agg(
    gasto_promedio        = ('gasto_total',             'mean'),
    satisfaccion_promedio = ('puntuacion_satisfaccion', 'mean'),
    descuento_promedio    = ('descuento_aplicado',      'mean')
).round(2)

print("\n" + "-"*50)
print("GASTO Y SATISFACCION POR TIPO DE CLIENTE")
print("-"*50)
display(satisfaccion_gasto)

cliente_valioso = satisfaccion_gasto['gasto_promedio'].idxmax()
sat = satisfaccion_gasto.loc[cliente_valioso, 'satisfaccion_promedio']

print("\n" + "-"*50)
print("RESULTADO:")
print(f"Segmento con mayor gasto promedio: [{cliente_valioso}]")
print(f"Satisfaccion promedio de este segmento: {sat}/5")

if sat < 3.5:
    print("  ADVERTENCIA: alta facturacion pero baja satisfaccion.")
    print("  Decision: revisar experiencia post-compra para reducir riesgo de fuga.")
else:
    print("  Decision: reforzar programa de fidelizacion para este segmento.")
print("-"*50)

# Grafico con colores por tipo de cliente y grid
plt.figure(figsize=(10,5))

colores = [get_color_cliente(c) for c in satisfaccion_gasto.index]
satisfaccion_gasto['gasto_promedio'].sort_values().plot(kind='barh', color=colores)

plt.title('Gasto Promedio por Segmento de Cliente', fontsize=14, fontweight='bold')
plt.xlabel('Gasto Promedio (euros)', fontsize=11)
plt.ylabel('Tipo de Cliente', fontsize=11)
plt.grid(axis='x', alpha=0.3, linestyle='--', color='#cccccc')

for i, v in enumerate(satisfaccion_gasto['gasto_promedio'].sort_values()):
    plt.text(v, i, f' {v:,.0f} €', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Insight 3 — Eficiencia del descuento por segmento

**Pregunta:** ¿Los descuentos se están aplicando en los segmentos que más valor generan?

```python
# Calculamos el descuento real en euros
df['descuento_euros'] = df['descuento_aplicado'] * df['gasto_total']

# Agrupamos por canal y tipo de cliente
eficiencia_descuento = df.groupby(['canal', 'tipo_cliente']).agg(
    descuento_euros = ('descuento_euros', 'mean'),
    gasto_euros = ('gasto_total', 'mean'),
    n = ('importe', 'count')
).round(2)

# Mostramos la tabla
print("\n" + "-"*60)
print("EFICIENCIA DE DESCUENTOS POR SEGMENTO")
print("-"*60)
display(eficiencia_descuento)

# Buscamos el segmento con mas descuento y el que mas gasta
mas_descuento = eficiencia_descuento['descuento_euros'].idxmax()
mas_gasto = eficiencia_descuento['gasto_euros'].idxmax()

print("\n" + "-"*60)
print("RESULTADO:")
print(f"Segmento que recibe mas descuento: {mas_descuento}")
print(f"Segmento que mas gasta: {mas_gasto}")

if mas_descuento != mas_gasto:
    print("\tPROBLEMA: El descuento no va al que mas gasta.")
    print("\tSOLUCION: Mover promociones al segmento que si genera valor.")
else:
    print("\tOK: Los descuentos estan bien asignados.")
print("-"*60)

# ============================================================================
# GRAFICO SIMPLE: comparamos lado a lado
# ============================================================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14,5))

segmentos = [f"{c[0]}-{c[1]}" for c in eficiencia_descuento.index]

# Grafico 1: Quien recibe mas descuento?
colores1 = [COLOR_RECURRENTE if c[1]=='Recurrente' else COLOR_NUEVO for c in eficiencia_descuento.index]
barras1 = ax1.barh(segmentos, eficiencia_descuento['descuento_euros'], color=colores1)
ax1.set_title('Descuento promedio por segmento', fontsize=12, fontweight='bold')
ax1.set_xlabel('Descuento (euros)')
ax1.grid(axis='x', alpha=0.3)

for i, v in enumerate(eficiencia_descuento['descuento_euros'].values):
    ax1.text(v + 0.5, i, f'{v:.1f} €', va='center', fontsize=9, fontweight='bold')

# Grafico 2: Quien gasta mas?
colores2 = [COLOR_TIENDA if c[0]=='Tienda' else COLOR_ONLINE for c in eficiencia_descuento.index]
barras2 = ax2.barh(segmentos, eficiencia_descuento['gasto_euros'], color=colores2)
ax2.set_title('Gasto promedio por segmento', fontsize=12, fontweight='bold')
ax2.set_xlabel('Gasto (euros)')
ax2.grid(axis='x', alpha=0.3)

for i, v in enumerate(eficiencia_descuento['gasto_euros'].values):
    ax2.text(v + 2, i, f'{v:.0f} €', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

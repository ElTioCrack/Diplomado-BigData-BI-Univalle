```python
# ============================================
# NOTEBOOK 1: GRAFICOS HONESTOS
# PRESUPUESTO DE MARKETING POR CANAL 2025
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Datos
data = {
    "Canal": ["Redes Sociales", "Google Ads", "Email", "TV", "Eventos", "Influencers"],
    "Q1": [45, 80, 15, 120, 30, 10],
    "Q2": [52, 75, 18, 110, 45, 25],
    "Q3": [48, 82, 20, 95, 25, 35],
    "Q4": [60, 90, 22, 85, 35, 50],
    "Total": [205, 327, 75, 410, 135, 120],
}

df = pd.DataFrame(data)

# ============================================
# GRAFICO 1: BARRAS AGRUPADAS POR TRIMESTRE
# ============================================

fig, ax = plt.subplots(figsize=(12, 6))

x = np.arange(len(df["Canal"]))
width = 0.2
trimestres = ["Q1", "Q2", "Q3", "Q4"]
colores = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]

for i, (trim, color) in enumerate(zip(trimestres, colores)):
    ax.bar(x + i * width, df[trim], width, label=trim, color=color)

ax.set_xlabel("Canal", fontsize=12)
ax.set_ylabel("Presupuesto (miles USD)", fontsize=12)
ax.set_title(
    "Presupuesto de Marketing por Canal y Trimestre", fontsize=14, fontweight="bold"
)
ax.set_xticks(x + width * 1.5)
ax.set_xticklabels(df["Canal"], rotation=45, ha="right")
ax.set_ylim(0, 130)
ax.legend()
ax.grid(axis="y", alpha=0.3)

# Fuente
# ax.text(
#     0.02,
#     -0.12,
#     "Fuente: Direccion de Marketing - Plan Anual 2025",
#     transform=ax.transAxes,
#     fontsize=9,
#     style="italic",
# )

plt.tight_layout()
plt.show()

# CONCLUSION GRAFICO 1
print("CONCLUSION:")
print("TV es el canal con mayor inversion, especialmente en Q1 con 120 mil USD.")
print(
    "Google Ads mantiene una inversion constante alrededor de 80 mil USD por trimestre."
)
print("Influencers muestra un crecimiento sostenido de 10 a 50 mil USD durante el año.")
print("Email es el canal con menor presupuesto, nunca superando los 22 mil USD.")

# ============================================
# GRAFICO 2: BARRAS HORIZONTALES - TOTAL POR CANAL
# ============================================

fig, ax = plt.subplots(figsize=(10, 6))

df_sorted = df.sort_values("Total", ascending=True)
bars = ax.barh(df_sorted["Canal"], df_sorted["Total"], color="steelblue")

ax.set_xlabel("Presupuesto Total (miles USD)", fontsize=12)
ax.set_ylabel("Canal", fontsize=12)
ax.set_title(
    "Presupuesto Total de Marketing por Canal (Anual)", fontsize=14, fontweight="bold"
)
ax.set_xlim(0, 450)
ax.grid(axis="x", alpha=0.3)

for bar, val in zip(bars, df_sorted["Total"]):
    ax.text(
        bar.get_width() + 5,
        bar.get_y() + bar.get_height() / 2,
        f"{val}",
        ha="left",
        va="center",
        fontweight="bold",
    )

# Fuente
ax.text(
    0.02,
    -0.12,
    "Fuente: Direccion de Marketing - Plan Anual 2025",
    transform=ax.transAxes,
    fontsize=9,
    style="italic",
)

plt.tight_layout()
plt.show()

# CONCLUSION GRAFICO 2
print("CONCLUSION:")
print(
    "TV lidera claramente con 410 mil USD, mas del doble que Google Ads (327 mil USD)."
)
print(
    "Redes Sociales (205 mil USD) y Eventos (135 mil USD) son el tercer y cuarto canal."
)
print("Email es el canal con menor presupuesto total (75 mil USD).")
print(
    "Influencers totaliza 120 mil USD, superando a Email pero lejos de los canales principales."
)

# ============================================
# GRAFICO 3: LINEAS DE TENDENCIA POR TRIMESTRE
# ============================================

fig, ax = plt.subplots(figsize=(12, 6))

trimestres = ["Q1", "Q2", "Q3", "Q4"]
canales_destacados = ["TV", "Google Ads", "Influencers", "Redes Sociales"]

for canal in canales_destacados:
    valores = df[df["Canal"] == canal][trimestres].values[0]
    ax.plot(trimestres, valores, marker="o", linewidth=2, label=canal)

ax.set_xlabel("Trimestre", fontsize=12)
ax.set_ylabel("Presupuesto (miles USD)", fontsize=12)
ax.set_title(
    "Evolucion Trimestral del Presupuesto por Canal", fontsize=14, fontweight="bold"
)
ax.set_ylim(0, 130)
ax.legend()
ax.grid(True, alpha=0.3)

# Fuente
ax.text(
    0.02,
    -0.12,
    "Fuente: Direccion de Marketing - Plan Anual 2025",
    transform=ax.transAxes,
    fontsize=9,
    style="italic",
)

plt.tight_layout()
plt.show()

# CONCLUSION GRAFICO 3
print("CONCLUSION:")
print("TV muestra una tendencia decreciente: de 120 a 85 mil USD durante el año.")
print("Influencers es el unico canal con crecimiento constante: +40 mil USD en total.")
print("Google Ads se mantiene estable, con un ligero pico en Q3 (82 mil USD).")
print("Redes Sociales crece moderadamente de 45 a 60 mil USD.")
```

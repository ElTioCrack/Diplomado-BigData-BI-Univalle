```python
# ============================================
# NOTEBOOK 1: GRAFICO HONESTO
# PRESUPUESTO DE MARKETING POR CANAL 2025
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Datos
data = {
    'Canal': ['Redes Sociales', 'Google Ads', 'Email', 'TV', 'Eventos', 'Influencers'],
    'Q1': [45, 80, 15, 120, 30, 10],
    'Q2': [52, 75, 18, 110, 45, 25],
    'Q3': [48, 82, 20, 95, 25, 35],
    'Q4': [60, 90, 22, 85, 35, 50],
    'Total': [205, 327, 75, 410, 135, 120]
}

df = pd.DataFrame(data)

# Calcular porcentajes sobre el total general
total_general = df['Total'].sum()
df['Porcentaje'] = (df['Total'] / total_general) * 100

# Ordenar de mayor a menor
df_sorted = df.sort_values('Total', ascending=True)

fig, ax = plt.subplots(figsize=(12, 7))

# Crear barras horizontales
bars = ax.barh(df_sorted['Canal'], df_sorted['Porcentaje'], color='steelblue', edgecolor='navy', linewidth=1)

ax.set_xlabel('Porcentaje del Presupuesto Total (%)', fontsize=12)
ax.set_ylabel('Canal', fontsize=12)
ax.set_title('Distribucion del Presupuesto de Marketing por Canal (Anual)', fontsize=14, fontweight='bold')
ax.set_xlim(0, 35)
ax.grid(axis='x', alpha=0.3)

# Mostrar ambos: porcentaje Y valor numerico
for bar, pct, total in zip(bars, df_sorted['Porcentaje'], df_sorted['Total']):
    ax.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height()/2, 
            f'{pct:.1f}% ({total} mil USD)', 
            ha='left', va='center', fontweight='bold', fontsize=10)

# Fuente
ax.text(0.02, -0.12, 'Fuente: Direccion de Marketing - Plan Anual 2025', 
        transform=ax.transAxes, fontsize=9, style='italic')

plt.tight_layout()
plt.show()

print("CONCLUSION:")
print(f"TV concentra el {(df[df['Canal']=='TV']['Porcentaje'].values[0]):.1f}% del presupuesto total (410 mil USD).")
print(f"Google Ads le sigue con {(df[df['Canal']=='Google Ads']['Porcentaje'].values[0]):.1f}% (327 mil USD).")
print(f"Influencers y Eventos representan {(df[df['Canal']=='Influencers']['Porcentaje'].values[0] + df[df['Canal']=='Eventos']['Porcentaje'].values[0]):.1f}% combinados.")
print(f"Email es el canal con menor asignacion: {df[df['Canal']=='Email']['Porcentaje'].values[0]:.1f}% (75 mil USD).")
```

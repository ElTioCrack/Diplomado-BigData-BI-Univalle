```python
# ============================================
# NOTEBOOK 2: GRAFICO ENGANOSO
# PRESUPUESTO DE MARKETING POR CANAL 2025
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# MISMOS DATOS que el honesto
data = {
    'Canal': ['Redes Sociales', 'Google Ads', 'Email', 'TV', 'Eventos', 'Influencers'],
    'Total': [205, 327, 75, 410, 135, 120]
}

df = pd.DataFrame(data)

# Calcular porcentajes (los mismos del honesto)
total_general = df['Total'].sum()
df['Porcentaje'] = (df['Total'] / total_general) * 100

# ENGANO 1: Reordenar para que Influencers aparezca primero (mas visible)
# En lugar de ordenar por tamaño, ordenamos para manipular percepcion
orden_enganoso = ['Influencers', 'Eventos', 'Redes Sociales', 'Google Ads', 'TV', 'Email']
df_enganoso = df.set_index('Canal').loc[orden_enganoso].reset_index()

valores = df_enganoso['Total'].values
canales = df_enganoso['Canal'].values
porcentajes = df_enganoso['Porcentaje'].values

# ENGANO 2: Colores - verde brillante para Influencers, gris opaco para TV
colores = []
for canal in canales:
    if canal == 'Influencers':
        colores.append('#2ecc71')  # Verde brillante (destaca)
    elif canal == 'TV':
        colores.append('#95a5a6')  # Gris opaco (resta importancia)
    elif canal == 'Google Ads':
        colores.append('#3498db')
    elif canal == 'Redes Sociales':
        colores.append('#9b59b6')
    elif canal == 'Eventos':
        colores.append('#f39c12')
    else:
        colores.append('#e74c3c')

# ENGANO 3: Explode - solo separamos a Influencers
explode = [0.08 if canal == 'Influencers' else 0 for canal in canales]

# ENGANO 4: Torta 3D con shadow
fig, ax = plt.subplots(figsize=(10, 8))

wedges, texts, autotexts = ax.pie(
    valores,
    labels=canales,
    autopct=lambda pct: f'{pct:.1f}%',
    explode=explode,
    colors=colores,
    startangle=0,  # Comienza con Influencers arriba
    shadow=True,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2},
    textprops={'fontsize': 11, 'fontweight': 'bold'}
)

# Formato de textos
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')
    autotext.set_fontsize(11)
    autotext.set_bbox(dict(boxstyle='round,pad=0.2', facecolor='black', alpha=0.5))

# Titulo ENGANOSO: igual que el honesto pero con torta 3D
ax.set_title('Distribucion del Presupuesto de Marketing por Canal (Anual)\nAnalisis 2025', 
             fontsize=14, fontweight='bold', pad=20, color='darkgreen')

# Fuente
ax.text(0.02, -0.12, 'Fuente: Direccion de Marketing - Plan Anual 2025', 
        transform=ax.transAxes, fontsize=9, style='italic', alpha=0.7)

plt.tight_layout()
plt.show()

print("="*60)
print("CONCLUSION DEL GRAFICO ENGANOSO")
print("="*60)
print("Influencers se posiciona como un canal estrategico fundamental,")
print("con una participacion destacada en el mix de marketing.")
print("Los canales tradicionales como TV y Google Ads, a pesar de su")
print("inversion historica, muestran una porcion ajustada que sugiere")
print("una migracion hacia nuevas estrategias digitales.")
print("Email complementa la estrategia con una asignacion eficiente.")
```
